# Shoulder Injury Risk Assessment

```python
from smash_analysis import (extract_kinematic_metrics, 
                            find_smashes, assess_smashes, 
                            plot_smashes_kinematics, 
                            overlay_kinematic_assessment)
from pathlib import Path
```

```python
# Example usage:
input_video = 'fixtures/strokes.mp4'

file_path = Path(input_video)
file_stem = file_path.stem

output_dir = "../test_output"
img_output_dir = output_dir
output_video = f"{output_dir}/{file_stem}_analyzed{file_path.suffix}"

Path(output_dir).mkdir(parents=True, exist_ok=True)
```

## Kinematic Metrics Extraction

```python
kin_metrics, fps = extract_kinematic_metrics(input_video)
```

## Smash Assessment

```python
smashes = find_smashes(kin_metrics, fps)
max_critical_dec = max([abs(s.critical_deceleration) for s in smashes] + [0])

assessments = assess_smashes(kin_metrics=kin_metrics,
                                smashes=smashes,
                                max_critical_deceleration=max_critical_dec)
```

## Visualization

### Graph

```python
plot_smashes_kinematics(metrics=kin_metrics,
                           fps=fps,
                           smashes=smashes,
                           output_dir=img_output_dir)
```

### Overlaying Video

```python
overlay_kinematic_assessment(
    input_video_path=input_video,
    output_video_path=output_video,
    metrics=kin_metrics,
    smashes=smashes,
    assessments=assessments,
    scale_linear=0.2,
    scale_ang_accel=0.05,
    show_labels=True,
    # pixels_per_meter=120
)
```
